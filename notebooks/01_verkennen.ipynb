## Na het draaien van `python -m src.bouw_data`
Daarna staat er een bestand met de afstand tot water per huis.

```python
# 6. eerste blik: prijs per m2 per afstandsgroep, voor water en voor groen
df = laden.laad_schoon()
water = analyse.mediaan_per_groep(df, "groep_water")
groen = analyse.mediaan_per_groep(df, "groep_groen")
water["laag"] = "water"
groen["laag"] = "groen"
tabel = pd.concat([water, groen])
tabel
```

```python
# 7. dezelfde tabel als grafiek
fig = px.bar(tabel, x="afstandsgroep", y="median", color="laag", barmode="group",
             labels={"afstandsgroep": "Afstand", "median": "Mediaan prijs per m2 (dollar)"})
fig.show()
```
